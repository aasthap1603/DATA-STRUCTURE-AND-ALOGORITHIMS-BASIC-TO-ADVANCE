```python
# IF CONDITION
age = 160

if age > 120:
    print("Invalid age found")
# Use if when you want to do something only when a condition is true, such as detecting missing or invalid data
```

```python
# IF  else condtion 
# real world applications ( data science or AI/ML )
# 1. fraud detection
amount = 50000
country = "foreign"

if amount > 20000 and country == "foreign":
    print("Flag transaction as suspicious")
else:
    print("Transaction approved")

# Example: if a credit card transaction is unusually large and from a new location, the system may alert the user or freeze the card.




```

```python
# 2. Data cleaning
age = None
if age is None:
    age = 0
    print("Missing age replaced")
else:
    print("Age is available")    
#Example: missing values in a dataset can be replaced with default values before training a model.
```

```python
# 3. Spam detection rules
subject = "Win a free iPhone"
sender = "unknown@example.com"

if "free" in subject.lower() or sender == "unknown@example.com":
    print("Mark as spam")
else:
    print("Inbox")
#Example: if an email contains suspicious words like “free” or “winner,” it may be filtered.
```

```python
# IF- ELSE - IF CONDITION 
probability = 0.82

if probability >= 0.9:
    print("High confidence")
elif probability >= 0.7:
    print("Medium confidence")
else:
    print("Low confidence")
#Example: a machine learning model might classify predictions as high, medium, or low confidence based on the probability score.
# REAL WORLD APPLICATIONS 
# Categorizing customers into segments.
# Assigning risk levels in fraud detection.
# Labeling sentiment as positive, neutral, or negative.  
```

```python
# SWITCH / MATCH CASE 
model_type = "regression"

match model_type:
    case "classification":
        print("Use accuracy")
    case "regression":
        print("Use MAE or RMSE")
    case "clustering":
        print("Use silhouette score")
    case _:
        print("Unknown model type")
# example : Choosing the right evaluation metric.
# Selecting preprocessing steps based on data type.
# Routing different model pipelines.
```

```python
# ternary operator 
score = 45
result = "Pass" if score >= 35 else "Fail"
print(result)
# Assigning “Yes/No” labels.
# Creating quick flags in dataframes.
# Setting default values during preprocessing.
```
